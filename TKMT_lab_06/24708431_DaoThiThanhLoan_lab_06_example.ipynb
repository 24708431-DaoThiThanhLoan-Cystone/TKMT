```python
import numpy as np
from scipy.stats import t

from scipy.stats import norm
```

```python
# a. Ước lượng khối lượng trung bình các quả cam với độ tin cậy 95%
#bài này tính tb, dùng bảng t (vì ko có độ lệch chuẩn)

x = np.array([32, 33, 34, 35, 36, 37, 38, 39, 40])
f = np.array([2, 3, 15, 26, 28, 6, 8, 8, 4])
n = f.sum()# Tổng số quả cam, nhưng đề cho 100 rồi thì có thể thế 100 vào

# Trung bình mẫu
tb = (x * f).sum() / n

# Độ lệch chuẩn mẫu
dlc = np.sqrt((f * (x - tb)**2).sum() / (n - 1))

# Khoảng tin cậy 95%
dotincay = 0.95
alpha = 1 - dotincay
t_alpha_chia2 = t.isf(alpha / 2, df=n - 1)

saiso = t_alpha_chia2 * dlc / np.sqrt(n)

print('Số quả cam:', n)
print('Trung bình mẫu:', round(tb, 3))
print('Độ lệch chuẩn mẫu:', round(dlc, 3))
print('Sai số:', round(saiso, 3))
print('Khoảng tin cậy 95%:',
      round(tb - saiso, 3),
      round(tb + saiso, 3))
```

```python
#b. cam có khối lượng dưới 34g được coi là cam loại 2. Tìm khoảng ước lượng cho tỷ lệ loại 2 với độ tin cậy 90%
#bài này tính tỷ lệ, dùng z

n = 100 #tổng
so_cam_loai2 = 2 + 3 #dưới 34 g chỉ gồm cam 32 g và 33 g => 2 + 3
# so_cam_loai2 = f[0] + f[1] #hoặc code như này

# Tỷ lệ mẫu
p_hat = so_cam_loai2 / n

# Độ tin cậy 90%
dotincay = 0.90
alpha_chia_2 = (1 - dotincay) / 2
z = norm.isf(alpha_chia_2)

# Sai số
saiso = z * np.sqrt(p_hat * (1-p_hat) / n)

print('Số cam loại 2:', so_cam_loai2)
print('Tỷ lệ mẫu:', p_hat)
print('Sai số:', round(saiso, 4))
print('Khoảng tin cậy 90%:',
      round(p_hat - saiso, 3),
      round(p_hat + saiso, 3))
```
