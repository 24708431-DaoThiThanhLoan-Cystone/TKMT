## <u>Dạng 1 — Ước lượng tỷ lệ (Bài 1, 2, 3)</u>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
```

## <u>Bài 1</u>

Green M&M Candies liên quan đến Dataset 18 trong file excel. Tìm tỉ lệ mẫu của M&M có màu xanh lá cây (green). Sử dụng kết quả để xây dựng 1 ước lượng khoảng tin cậy 95% của % quần thể M&M có màu xanh.

```python
df = pd.read_excel("./data/18_M_M.xls")
df.head()
```

```python
# Tổng số M&M màu xanh
# n = df['Green'].count() #cái này hnhu sai
n = df['Green'].sum()

# Tổng số M&M
N = df.sum().sum()

# Tỷ lệ mẫu
tyle_green = n / N

print('Số M&M màu xanh:', n)
print('Tổng số M&M:', N)
print('Tỷ lệ mẫu:', tyle_green)
```

```python
# tính sai số 
dotincay = 0.95
alpha_chia_2 = (1 - dotincay) / 2 #phần xác suất còn lại.
Z_alpha_chia2 = norm.isf(alpha_chia_2)

saiso = Z_alpha_chia2 * np.sqrt(tyle_green * (1 - tyle_green) / N)

print('Sai số ước lượng:', saiso.round(4))

#tính khoảng tin cậy
print('Khoảng tin cậy:')
print(
    (tyle_green - saiso).round(4),
    (tyle_green + saiso).round(4)
)
```

## <u>Bài 2</u>

Freshman Weight Gain liên quan đến Dataset 3 trong file excel

a. Dựa vào kết quả của mẫu, tìm ước lượng điểm tốt nhất của tỉ lệ phần trăm các sinh viên cao đẳng tăng cân trong năm thứ 1.

b. Xây dựng ước lượng khoảng tin cậy 95% về tỉ lệ phần trăm các sinh viên cao đẳng tăng cân trong năm thứ 1.


```python
df2 = pd.read_excel('./data/03_FRESH15.xls')
df2.head()
```

```python
#a. Ước lượng điểm (lấy dữ liệu cân nặng tháng 9 và tháng 4 từ bảng df2)
data_t9 = df2['WTSEP']
data_t4 = df2['WTAPR']

# Những sinh viên tăng cân
data_tang = df2.loc[data_t9 > data_t4] #Dòng này chọn những sinh viên có cân nặng tháng 9 lớn hơn tháng 4


# tính tỷ lệ mẫu
n = data_tang['WTSEP'].count()
N = data_t9.count() #lấy t9 hay t4 đều được

tyle = n / N

print('Tổng số sinh viên:', N)
print('Số sinh viên tăng cân:', n)
print('Tỷ lệ tăng cân:', tyle)
```

```python
#b. Khoảng tin cậy 95%
dotincay = 0.95
alpha_chia_2 = (1 - dotincay) / 2
Z_alpha_chia2 = norm.isf(alpha_chia_2)

saiso = Z_alpha_chia2 * np.sqrt(tyle * (1 - tyle) / N)

print('Sai số ước lượng:', saiso.round(4))

print('Khoảng tin cậy:')
print(
    (tyle - saiso).round(4),
    (tyle + saiso).round(4)
)
```

## <u>Bài 4</u>

Tổng số tiền phim: liên quan đến Dataset 9 trong file excel. Xây dựng ước tính khoảng thời gian tin cậy 95% của tổng số tiền trung bình cho quần thể của tất cả các phim. Giả định rằng độ lệch chuẩn của quần thể được biết là 100 triệu đô la. 

```python
df4 = pd.read_excel('./data/09_MOVIES.xls')
df4.head()
```

```python
#coi cái bên dưới cho dễ hiểu
#tổng số tiền phim
data = df4['Gross'] #gross là doanh thu phim
N = data.count() #Số lượng phim
sigma = 100
tb = data.mean() #Trung bình mẫu

sigma_moi = sigma / np.sqrt(N) #tính sai số chuẩn của trung bình.
Z_alpha_chia2 = norm.isf(0.025) #lấy Z cho mức tin cậy 95%, xấp xỉ 1.96.
saiso = Z_alpha_chia2 * sigma_moi #Sai số ước lượng

print('Số lượng phim:', N)
print('Trung bình mẫu:', round(tb,4))
print('Sai số ước lượng:', saiso.round(4))
print('Khoảng ước lượng:')
print(
    (tb - saiso).round(4),
    (tb + saiso).round(4)
)
```

```python
#dễ hiểu hơn
# Lấy dữ liệu doanh thu phim
data = df4['Gross']

# Số lượng phim và doanh thu trung bình
N = data.count()
tb = data.mean()

# Độ lệch chuẩn quần thể đề bài cho
sigma = 100

# Tính giá trị Z với độ tin cậy 95%
dotincay = 0.95
alpha_chia_2 = (1 - dotincay) / 2
Z = norm.isf(alpha_chia_2)

# Tính sai số ước lượng
saiso = Z * sigma / np.sqrt(N)

# Tính khoảng tin cậy
can_duoi = tb - saiso
can_tren = tb + saiso

print('Số lượng phim:', N)
print('Trung bình mẫu:', round(tb, 4))
print('Sai số ước lượng:', round(saiso, 4))
print('Khoảng tin cậy 95%:', round(can_duoi, 4), round(can_tren, 4))
```

## <u>Bài 5</u>

Điểm đánh giá tín dụng FICO: liên quan đến Dataset 24 trong file excel. Xây dựng ước lượng khoảng tin cậy 99% của điểm FICO trung bình cho quần thể. Giả sử độ lệch chuẩn của quần thể là 92,2.

```python
col_names = ['Diem']
df5 = pd.read_excel(
    './data/24_FICO.xls',
    names=col_names,
    header=None #file Excel không có dòng tiêu đề cột
)
df5.head()
```

```python
#khó hiểu
data = df5['Diem']
N = data.count()
tb = data.mean()
sigma = 92.2
sigma_moi = sigma / np.sqrt(N)
Z_alpha_chia2 = norm.isf(0.01 / 2)
saiso = Z_alpha_chia2 * sigma_moi

print('Số lượng:', N)
print('Trung bình mẫu:', tb)
print('Độ lệch chuẩn quần thể:', sigma)
print('Z:', Z_alpha_chia2.round(4))
print('Sai số ước lượng:', saiso.round(4))
print('Khoảng tin cậy 99%:')
print(
    (tb - saiso).round(4),
    (tb + saiso).round(4)
)
```

```python
#dễ hiểu hơn
# Lấy dữ liệu điểm FICO
data = df5['Diem']

# Số lượng và điểm trung bình
N = data.count()
tb = data.mean()

# Độ lệch chuẩn quần thể đề bài cho
sigma = 92.2

# Tính giá trị Z với độ tin cậy 99%
dotincay = 0.99
alpha_chia_2 = (1 - dotincay) / 2
Z = norm.isf(alpha_chia_2)

# Tính sai số ước lượng
saiso = Z * sigma / np.sqrt(N)

# Tính khoảng tin cậy
can_duoi = tb - saiso
can_tren = tb + saiso

print('Số lượng:', N)
print('Trung bình mẫu:', round(tb, 4))
print('Sai số ước lượng:', round(saiso, 4))
print('Khoảng tin cậy 99%:', round(can_duoi, 4), round(can_tren, 4))
```
